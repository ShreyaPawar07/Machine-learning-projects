```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
docs = ['office meet tomorrow' , 'office furniture tomorrow' , 'meet me in the office' , 'tomorrow is meet day']
analyzer = TfidfVectorizer()

word_score= analyzer.fit_transform(docs)

print(word_score.toarray())
```

```python
query = 'furniture'
query_scores = analyzer.transform([query])
print(query_scores.toarray())
```

```python
# Using BM25 
from rank_bm25 import BM25Okapi
```

```python
bm25 = BM25Okapi([doc.split() for doc in docs])
wscores = bm25.get_scores("office meet")
print(wscores)
```
